# Tarea 8.2: Separación óptima entre puntos


Tomamos $f(x) = \sin(x)$ en $[0, 4]$ con datos ruidosos $f_i = \sin(x_i) + \delta_i$, donde $|\delta_i| \le \Delta$ y $\Delta = 0.001$.

Usamos la fórmula de tres puntos $f'(x_i) \approx \dfrac{f_{i+1} - f_{i-1}}{2h}$. El error total tiene dos partes:

- **Truncamiento:** $\dfrac{M}{6}h^2$, con $M = \max|f'''| = 1$. Disminuye al bajar $h$.
- **Ruido:** $\dfrac{\Delta}{h}$. Aumenta al bajar $h$.

Entonces $E(h) = \dfrac{\Delta}{h} + \dfrac{M}{6}h^2$.

Luego

Derivando e igualando a cero:

$$-\frac{\Delta}{h^2} + \frac{M}{3}h = 0 \quad\Rightarrow\quad h_{\text{óptimo}} = \left(\frac{3\Delta}{M}\right)^{1/3} \approx 0.144$$

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

DELTA = 0.001   # amplitud del ruido
M = 1.0         # cota de |f'''| para f = sin

# Separación óptima teórica
h_opt = (3 * DELTA / M) ** (1/3)
print(f"h óptimo teórico = {h_opt:.4f}")
```

## Simulación

Para cada $h$ generamos datos con ruido, calculamos la derivada y medimos el error máximo frente a $\cos(x)$. Repetimos 200 veces y promediamos, porque el ruido es aleatorio.

```python
def error_promedio(h, n_rep=200):
    # Puntos separados por h en [0, 4]
    x = np.arange(0, 4 + h/2, h)
    exacta = np.cos(x)

    errores = []
    for _ in range(n_rep):
        datos = np.sin(x) + np.random.uniform(-DELTA, DELTA, len(x))
        # Derivada centrada en los puntos interiores
        aprox = (datos[2:] - datos[:-2]) / (2*h)
        # Comparamos sólo en los puntos interiores
        errores.append(np.max(np.abs(aprox - exacta[1:-1])))
    return np.mean(errores)

hs = np.linspace(0.05, 0.5, 46)
errores = np.array([error_promedio(h) for h in hs])

h_num = hs[np.argmin(errores)]
print(f"h con menor error (simulación) = {h_num:.3f}")
```

## Gráfica

```python
plt.figure(figsize=(7, 4))
plt.plot(hs, errores, 'o-', ms=3, label='Error promedio (simulación)')
plt.axvline(h_opt, color='red', linestyle='--', label=f'h óptimo teórico = {h_opt:.3f}')
plt.xlabel('separación h')
plt.ylabel('error máximo')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

Finalmente, reducir $h$ no siempre mejora la derivada: con $h$ muy pequeño el ruido domina, y con $h$ muy grande domina el error de truncamiento. La separación óptima teórica es $h \approx 0.144$ (unos 29 puntos en $[0, 4]$). La simulación ubica el mínimo en la misma zona, por lo que la fórmula analítica es una buena guía.
