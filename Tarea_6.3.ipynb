# Tarea 6.3: Bisección con criterio de parada relativo

**Enunciado:** Buscar un ejemplo de una función tal que el método de bisección con la condición $\frac{|p_n - p_{n-1}|}{|p_n|} < \epsilon$ converja siempre a un valor que **no** es un cero de la función.

**Propuesta:** Consideramos la función $f(x) = \frac{1}{x-1}$ en el intervalo $[0, 3]$. En los extremos hay cambio de signo, pero la función presenta una discontinuidad en $x=1$ y nunca se anula. De esta forma, la bisección se estabilizará en la discontinuidad.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Definimos la función con la discontinuidad
def f(x):
    return 1 / (x - 1)

def biseccion(f, a, b, eps=0.01, max_iter=100):
    fa = f(a)
    fb = f(b)
    if fa * fb > 0:
        raise ValueError("No hay cambio de signo en el intervalo")
    
    p_prev = None
    historial = []
    
    for n in range(1, max_iter + 1):
        p = (a + b) / 2
        
        # Calculamos el error relativo a partir de la segunda iteración
        if p_prev is not None:
            error_rel = abs(p - p_prev) / abs(p) if p != 0 else np.inf
            historial.append((n, p, error_rel))
            
            if error_rel < eps:
                break
        else:
            historial.append((n, p, None))
        
        # Actualizamos el intervalo según el signo
        fp = f(p)
        if fa * fp < 0:
            b = p
        else:
            a = p
            fa = fp
        p_prev = p
        
    return historial

# Ejecutamos el método
a, b = 0, 3
eps = 0.01
historial = biseccion(f, a, b, eps)

# Mostramos los resultados en una tabla
print(f"{'n':>3} | {'p_n':>10} | {'Error Relativo':>15}")
print("-" * 35)
for n, p, err in historial:
    if err is None:
        print(f"{n:>3} | {p:>10.6f} | {'-':>15}")
    else:
        print(f"{n:>3} | {p:>10.6f} | {err:>15.6f}")

p_final = historial[-1][1]
print(f"\nEl método se detuvo en p = {p_final:.6f}")
print(f"Evaluando f(p) obtenemos = {f(p_final):.4f} (No es cero)")
```

```python
# Graficamos para visualizar cómo los puntos se acercan a la asíntota
x = np.linspace(0, 3, 400)
x = x[np.abs(x - 1) > 0.05] # Evitamos la división por cero en la gráfica

plt.figure(figsize=(6, 4))
plt.plot(x, f(x), label=r"$f(x)=1/(x-1)$")
plt.axvline(1, color='red', linestyle='--', label="Discontinuidad x=1")
plt.scatter([p for _, p, _ in historial], [f(p) for _, p, _ in historial], color='orange', zorder=5, label="Iterados")
plt.ylim(-10, 10)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Conclusión
Observamos que el método se detuvo porque los valores de $p_n$ y $p_{n-1}$ se volvieron muy cercanos (se cumplió el criterio relativo). Sin embargo, el valor al que llegó es $x \approx 1$, donde la función tiene una asíntota y **no** un cero. 

En sencillas palabroas, comprobamos que el criterio de parada solo mide si el punto se estabilizó, pero no asegura que hayamos encontrado la raíz real.
